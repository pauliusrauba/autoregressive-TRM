```python
import pandas as pd
from pathlib import Path

# =============================================================================
# Load all exported experiment data
# =============================================================================
DATA_DIR = Path("data")

# Define all experiment tags
EXPERIMENT_TAGS = [
    "lengthgen_train",
    "compute_optimal",
    "compute_optimal-v2",
    "hl_structure",
    "trainlen_scaling",
    "paraminf_scaling",
]

# Load all data into dictionaries
summary_dfs = {}
history_dfs = {}

for tag in EXPERIMENT_TAGS:
    summary_path = DATA_DIR / f"{tag}_summary.csv"
    history_path = DATA_DIR / f"{tag}_history.csv"
    
    if summary_path.exists():
        summary_dfs[tag] = pd.read_csv(summary_path)
        print(f"✓ {tag}_summary: {len(summary_dfs[tag])} runs")
    else:
        print(f"✗ {tag}_summary: not found")
    
    if history_path.exists():
        history_dfs[tag] = pd.read_csv(history_path)
        print(f"✓ {tag}_history: {len(history_dfs[tag])} rows")
    else:
        print(f"✗ {tag}_history: not found")

# =============================================================================
# Quick access aliases (optional - for convenience)
# =============================================================================
# Summary dataframes
df_lengthgen_summary = summary_dfs.get("lengthgen_train")
df_compute_optimal_summary = summary_dfs.get("compute_optimal")
df_compute_optimal_v2_summary = summary_dfs.get("compute_optimal-v2")
df_hl_structure_summary = summary_dfs.get("hl_structure")
df_trainlen_summary = summary_dfs.get("trainlen_scaling")
df_paraminf_summary = summary_dfs.get("paraminf_scaling")

# History dataframes
df_lengthgen_history = history_dfs.get("lengthgen_train")
df_compute_optimal_history = history_dfs.get("compute_optimal")
df_compute_optimal_v2_history = history_dfs.get("compute_optimal-v2")
df_hl_structure_history = history_dfs.get("hl_structure")
df_trainlen_history = history_dfs.get("trainlen_scaling")
df_paraminf_history = history_dfs.get("paraminf_scaling")

# =============================================================================
# Print summary
# =============================================================================
print("\n" + "="*60)
print("DATA LOADED")
print("="*60)
print(f"Summary dataframes: {list(summary_dfs.keys())}")
print(f"History dataframes: {list(history_dfs.keys())}")
```
